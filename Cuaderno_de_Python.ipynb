```python
# Loops
seq_5 = 'ATGCTTGACTAGTACGTACGTAGCTAATGCCTAAATGGCTTGACTAGTACGTACGTAGCTGTACGTACGTAGCTAATGCTTGACTAGTACGTACGTAGCTAATGCTTGACTAGTACGTACGTAGCTA'
```

```python
#La funcion 'FOR' va a evaluar sobre cada posicion de la estructura
def ContenidoGC_v2(DNA):
  counter = 0
  for base in DNA:
    if base == 'C' or base == 'G':
      counter = counter + 1
  return counter/len(DNA)
```

```python
ContenidoGC_v2(seq_5)
```
